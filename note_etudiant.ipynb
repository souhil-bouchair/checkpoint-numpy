```python
import numpy as np

nb_etudiants = int(input("Nombre d'étudiants : "))
nb_matieres = int(input("Nombre de matières : "))

notes = np.zeros((nb_etudiants, nb_matieres))
noms = []

for i in range(nb_etudiants):
  nom = input(f"Nom de l'étudiant {i + 1} : ")
  noms.append(nom)
  for j in range(nb_matieres):
    notes[i, j] = float(input(f"Note de {nom} - matière {j + 1} : "))

totaux = np.sum(notes, axis=1)
pourcentages = (totaux / (nb_matieres * 100)) * 100

mentions = []
for p in pourcentages:
  if p >= 90:
    mentions.append("A+")
  elif p >= 80:
    mentions.append("A")
  elif p >= 70:
    mentions.append("B+")
  elif p >= 60:
    mentions.append("B")
  elif p >= 50:
    mentions.append("C")
  else:
    mentions.append("F")

print(f"\n{'Nom':<15} {'Total':<10} {'Pourcentage':<13} {'Mention'}")
print("-" * 48)
for i in range(nb_etudiants):
  print(
      f"{noms[i]:<15} {totaux[i]:<10.1f} {pourcentages[i]:<12.1f}% {mentions[i]}"
  )
```
